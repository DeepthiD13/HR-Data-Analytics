## Checking duplicate Employee IDs

Reason: We need to check whether Employee IDs are repeated in the dataset. Employee ID is used to identify individual employees, so repeated IDs may indicate duplicate employee records that require investigation.

```python

print("Total Employee Records:", len(df))

print("Unique Employee IDs:", df["Employee ID"].nunique())

print("Duplicate Employee IDs:", df["Employee ID"].duplicated().sum())
```

## Verifying duplicate employee records

Reason: Before removing duplicate records, we need to confirm whether the rows with repeated Employee IDs contain exactly the same information. This prevents us from accidentally deleting valid employee records that have different information.

```python

duplicate_id_records = df[df.duplicated(subset="Employee ID", keep=False)]

verification = (
    duplicate_id_records
    .groupby("Employee ID")
    .nunique(dropna=False)
    .max(axis=1)
    .eq(1)
)

print("Are all repeated Employee ID records identical?",
      verification.all())
```

## Removing duplicate records

Reason
Duplicate records can lead to inaccurate employee counts and misleading HR analysis. Since we have verified that the 150 repeated records are exact copies, we will remove the duplicate rows and retain one original record for each employee.

```python

df = df.drop_duplicates().copy()

print("Dataset after removing duplicates:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Remaining Duplicate Rows:", df.duplicated().sum())
print("Unique Employee IDs:", df["Employee ID"].nunique())
```

We have successfully removed 150 exact duplicate records without losing any unique employee IDs. The dataset now contains 3,000 unique employee records.

### Removing Unnecessary Spaces

Unnecessary leading and trailing spaces are removed from categorical text columns to maintain consistent category names and improve data quality.

```python
for column in categorical_columns:
    df[column] = df[column].str.strip()

print("Title values after removing spaces:")
print(df["Title"].unique())

print("\nDepartmentType values after removing spaces:")
print(df["DepartmentType"].unique())
```

### Verifying Categorical Data

The categorical columns are checked again to confirm that unnecessary leading and trailing spaces have been removed.

```python
for column in categorical_columns:
    space_values = df[column].dropna()

    inconsistent_values = space_values[
        space_values != space_values.str.strip()
    ]

    if len(inconsistent_values) > 0:
        print(column, "contains extra spaces")
    else:
        print(column, "is clean")
```

### Investigating Training Cost

The distribution of Training Cost is examined to identify unusually high or low values and potential outliers.

```python
print("Training Cost Summary:")
print(df["Training Cost"].describe())

print("\nLowest 10 Training Costs:")
print(df["Training Cost"].nsmallest(10))

print("\nHighest 10 Training Costs:")
print(df["Training Cost"].nlargest(10))
```

###

### Detecting Potential Outliers Using IQR

The Interquartile Range (IQR) method is used to identify potential outliers in the Training Cost column.

```python
Q1 = df["Training Cost"].quantile(0.25)
Q3 = df["Training Cost"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Training Cost"] < lower_bound) |
    (df["Training Cost"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Potential Outliers:", len(outliers))
```

###
*IQR calculation confirms that Training Cost has zero potential outliers.


*All Training Cost values fall within the calculated IQR boundaries.
